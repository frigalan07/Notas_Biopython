<Definición>Una base de datos es un conjunto de datos de varios tipos, organizados para optimizar su almacenamiento y agilizar su consulta. Por lo que existen interelaciones lógicas entre los datos para satisfacer los requerimientos de información de una organización.


1) **¿Qué es ENTREZ?** Es un sistema de recuperación de información de las bases de datos que estan vinculadas a NCBI  

Las E-utilities son una serie de interfaces de programación de aplicaciones (APIs)
 > API es un conjunto de reglas y definiciones que permite que diversos programas o sistemas intercambien información (comparten información sin necesariamente saber el funcionamiento interno de cada una)

que permiten acceder de manera programatica a las bases de datos. Estan diseñadas para ser accesibles a través de solicitudes HTTP(hipertext transfer protocol, que es una comunicación que se utiliza vía internet para pedir información a un servidor para pedir algún recurso)

Las principales E-utilities incluyen:

- ESearch: Para realizar búsquedas en una base de datos y recuperar identificadores de registros.
- ESummary: Para obtener resúmenes de registros basados en sus identificadores.
- EFetch: Para recuperar registros completos en formatos como XML, ASN.1, o texto plano.
- ELink: Para encontrar enlaces entre diferentes bases de datos dentro de NCBI.
- EInfo: Para obtener información sobre una base de datos específica, como los campos de búsqueda disponibles.
- EPost: Para cargar una lista de identificadores en un historial de búsqueda temporal en el servidor de NCBI.
- EGquery: Para realizar búsquedas globales en todas las bases de datos de NCBI a la vez.



**Einfo** 
Proporciona la cantidad de registro indexados en cada campo de una base de datos determinada, la fecha de la última actualización de la base de datos y los enlaces disponibles a otras bases de datos. TE PERMITE OBTENER METADATOS 

> El método Entrez.read(handle) toma ese flujo de datos (el "handle") y:

1) Lee los datos: Extrae los datos del flujo que has recibido de NCBI.
2) Parsea los datos: Convierte esos datos, que están en formato XML, en una estructura de datos de Python, como un diccionario, listas, cadenas, etc. Esto es muy útil porque te permite acceder a los datos de manera sencilla, usando los métodos y operadores estándar de Python.
3) Devuelve la estructura de datos: Devuelve un objeto de Python (generalmente un diccionario) que contiene todos los datos que venían en la respuesta del servidor NCBI.

````python

#1. Genera una consulta con Entrez.einfo a la base de datos de pubmed, para imprimir el nombre y la descripción de todos los FieldList

from Bio import Entrez 

#Especifico correo 
Entrez.email = "fridagh@lcg.unam.mx"

'''
**Utilizo handle con einfo*+
Handle será el objeto que contendrá la respuesta de la base de datos. 
Es el objeto que nos permitirá  manejar el flujo de datos que recibimos de la base de datos
Contiene la respuesta del servidor NCBI   
'''
handle = Entrez.einfo(db = "pubmed") 
'''
**Uso de record** 
<record> es la estructura de datos (objeto) que almacena la información obtenida de la consulta a NCBI 
Es comun llamar record a una variable que almacena una colección de datos estructurados que se obtienen de una 
fuente externa 
Contiene los datos que se han leído desde el objeto handle(contenido en bruto que arroja el servidor, generalmente un flujo de datos en XML) 
y se 
convierte en un objeto más estructurado y facil  de manipular, como diccionarios
'''
record = Entrez.read(handle)

handle.close()

print(type(record)) #Con esta línea de codigo yo puedo saber quue tipo de resultado me dará y podré saber como acceder a esos datos 

for field in record["DbInfo"]["FieldList"]:
    print(field["Name"] + ", " + field["Description"])

'''
Se accede de esta manera ya que estamos accediendo al primer diccionario, despues accedemos a FieldList que es una lista que adentro tiene ootros 
diccionario por eso usamos un iterador para acceder a todos los campos de la lista 
'''
````

**Efectch** 
Regresa registros en el formato indicado por el usuario de una lista de identificadores de entrada.
A diferencia de esummary, que te da un resumen conciso, efectch te da el contenido completo (como artículos completos, secuencias de DNA, etc ...)

**Detalles de Entrez.efetch**
1) **Base de datos (db):** Especifica la base de datos de la cual quieres recuperar los registros. Por ejemplo, "pubmed" para artículos científicos o "nucleotide" para secuencias de ADN.
2) **UIDs (id):** Es una lista o cadena de identificadores únicos (UIDs) que corresponden a los registros que quieres recuperar. Estos UIDs normalmente se obtienen usando esearch.
3) **Formato de retorno (retmode):**  Especifica el formato en el cual quieres recibir los datos. Por ejemplo:
- "xml": Devuelve los datos en formato XML, que puedes procesar con Entrez.read().
- "text": Devuelve los datos en texto plano.
- "html": Devuelve los datos en formato HTML.
- "json": Devuelve los datos en formato JSON (si es soportado por la base de datos).
4) **Tipo de registro (rettype):** Especifica el tipo de datos que deseas obtener. Por ejemplo, si estás accediendo a PubMed, puedes pedir el artículo en formato abstracto, completo (medline), entre otros. Si estás en una base de datos de secuencias, puedes especificar formatos como "fasta" o "gb" (GenBank).

````python
from Bio import Entrez, SeqIO #SeqIO se utiliza para leer y escribir secuencias en varios formatos (como fasta o GenBank)

handle = Entrez.efetch(db="nucleotide", id="HE805982", rettype="gb", retmode="text")
record = SeqIO.read(handle, "genbank")#Lectura e interpretacion del handle, devuelve un objeto del tipo 'SeqRecord' 
handle.close()

print(record)
#print(record.id) 
#print(record.description)

'''
Accedemos con un punto porque estamos accediendo a los atributos o métodos de ese objeto
En este caso accedemos a los atributos 
'''
````

````python
#Guardando en un archivo 

from Bio import Entrez, SeqIO

filename = "HE805982.gb" #Establecemos el nombre del archivo a generar, donde se guardara la secuencia descargada  
with Entrez.efetch(db = "nucleotide", id="HE805982", rettype="gb", retmode="text") as file: #Descarga de la secuencia 
    with open(filename, "w") as handle:  
        handle.write(file.read()) 

record = SeqIO.read("HE805982.gb", "genbank")
print(record)

'''
+ with ... as file:: Esto abre una conexión para descargar el archivo y lo guarda temporalmente en file.
La estructura with asegura que la conexión se cierre automáticamente después de que se complete el bloque de código.
+ with open(filename, "w") as handle:: Abre (o crea) un archivo con el nombre filename en modo de escritura ("w").
+ handle.write(file.read()): Escribe en el archivo todo el contenido leído de file (la secuencia descargada desde NCBI).

Lectura del archivo:
+ SeqIO.read: Se utiliza para leer la secuencia almacenada en el archivo HE805982.gb.
+ "genbank": Especifica que el formato del archivo es GenBank.
+ El resultado se almacena en la variable record, que es un objeto de tipo SeqRecord.
'''
````

**Elink**
Buscar información en otras bases de datos. 

Cuando buscamos más de un ID los resultados no te los va a regresar necesariamente en orden.

> Entrez.elink es una herramienta poderosa de NCBI que permite encontrar relaciones entre distintos registros de sus bases de datos. Por ejemplo, si tienes el ID de una proteína, puedes usar elink para encontrar el gen correspondiente, artículos relacionados, o incluso estructuras 3D relacionadas.

+ dbfrom: La base de datos de origen de los IDs proporcionados.
+ id: Los IDs específicos de la base de datos de origen que quieres vincular.
+ db: La base de datos de destino, donde quieres buscar registros relacionados.

El resultado es un conjunto de enlaces que muestran cómo los elementos en dbfrom (por ejemplo, proteínas) se relacionan con elementos en db (por ejemplo, genes).

````python
from Bio import Entrez 
from pprint import pprint

Entrez.email = "fridagh@lcg.unam.mx"

ids = "15718680, 157427902"
record = Entrez.read(Entrez.elink(dbfrom= "protein", id=ids,db="gene"))
print(type(record))
print(len(record))
print(type(record[0]))
print(record[0].keys())
pprint(record[0])
````

````python
'''
define una función llamada elink_multiple que genera una URL para realizar solicitudes a la API de NCBI Entrez, 
específicamente utilizando el servicio elink

mirror (str): La URL base para las solicitudes de Entrez (por defecto es la URL oficial de NCBI).
Retorna:
    handle: Un objeto de archivo que contiene la respuesta HTTP de la solicitud.

'''
# Función para generar la URL como en la documentación de ENTREZ

#Importación de librerias necesarias para realizar solicitudes HTTP y codificar parametros en una URL
from urllib.request import urlopen
from urllib.parse import urlencode

def elink_multiple(dbfrom, ids, db,mirror="https://eutils.ncbi.nlm.nih.gov/entrez/eutils"):
    # diccionario con los parametros necesarios para construir el URL
    parameters = {"dbfrom": dbfrom, "db":db, "id": ids, "tool":"biopython", "email":Entrez.email}
    # Creamos la URL
    #Codifica los parámetros en una cadena de consulta adecuada para una URL
    command = urlencode(parameters, doseq=True)
    #Construimos la URL completa utilizando la URL base (mirror) y los parámetros codificados 
    url = "%s/elink.fcgi?%s" % (mirror, command)

    #Abre la URL y devueleve un objeto de archivo que puede leerse para obtener la respuesta de NCBI
    handle = urlopen(url)
    return(handle)


pmids = ["15718680","157427902"] #Definimos IDs a buscar 
handle = elink_multiple(dbfrom="protein", ids=pmids, db="gene")
print(handle.url)

record = Entrez.read(handle)
handle.close()

print(type(record))
print(len(record))
print(type(record[0]))
print(record[1].keys())
print(record[1])
````

````python
"""
**OBTENER CITAS** 
Este código está diseñado para obtener citas (o referencias) de un artículo de PubMed a partir de su ID (PMID) 
y encontrar los artículos relacionados en PubMed Central (PMC)

parámetros 
+ dbfrom: La base de datos de origen desde donde se parte (por ejemplo, "pubmed").
+ db: La base de datos de destino hacia la cual se busca la relación (por ejemplo, "pmc").
+ LinkName: Especifica el tipo de enlace que se desea buscar. Este es un identificador para el tipo de relación entre las bases de datos (por ejemplo, "pubmed_pmc_refs" para encontrar referencias de PubMed en PMC).
+ from_uid: El ID del registro en la base de datos de origen que se usará para buscar enlaces.+ 
"""
from Bio import Entrez
from pprint import pprint

Entrez.email = "fridagh@lcg.unam.mx"
pmid = "32703847" #pubmed id


results = Entrez.read(Entrez.elink(dbfrom="pubmed", db="pmc",LinkName="pubmed_pmc_refs", from_uid=pmid))
# Guardemos links de PMC
pmc_ids = [link["Id"] for link in results[0]["LinkSetDb"][0]["Link"]]
"""
results[0]: Accede al primer (y en este caso, único) elemento de la lista de resultados.
results[0]["LinkSetDb"][0]["Link"]: Accede a la lista de enlaces dentro del resultado. Aquí, LinkSetDb es una clave que contiene los conjuntos de enlaces.
[link["Id"] for link in ...]: Utiliza una lista por comprensión para extraer los IDs de los artículos en PMC que están referenciados por el artículo de PubMed.
"""
print(pmc_ids)

#ahora partimos de PMC a pubmed
results2 = Entrez.read(Entrez.elink(dbfrom="pmc", db="pubmed", LinkName="pmc_refs_pubmed",id=",".join(pmc_ids)))
#guardamos links
pubmed_ids = [link["Id"] for link in results2[0]["LinkSetDb"][0]["Link"]]
print(pubmed_ids)
````
